```python
# Subsetting the wave datasets for mental health & social support data
b_mh_ss = b_indresp[['pidp', 'b_istrtdaty', 'b_scghq1_dv', 'b_scfrendany', 'b_scpundstnd', 'b_scprely', 'b_scpopenup', 'b_scpcritic', 'b_scpletdwn', 'b_scpannoy', 'b_screlany', 'b_scrundstnd', 'b_scrrely', 'b_scropenup', 'b_scrcritic', 'b_scrletdwn', 'b_scrannoy', 'b_scfundstnd', 'b_scfrely', 'b_scfopenup', 'b_scfcritic', 'b_scfletdwn', 'b_scfannoy']].copy() # wave 2
e_mh_ss = e_indresp[['pidp', 'e_istrtdaty', 'e_scghq1_dv', 'e_scfrendany', 'e_scpundstnd', 'e_scprely', 'e_scpopenup', 'e_scpcritic', 'e_scpletdwn', 'e_scpannoy', 'e_screlany', 'e_scrundstnd', 'e_scrrely', 'e_scropenup', 'e_scrcritic', 'e_scrletdwn', 'e_scrannoy', 'e_scfundstnd', 'e_scfrely', 'e_scfopenup', 'e_scfcritic', 'e_scfletdwn', 'e_scfannoy']].copy() # wave 5
g_mh = g_indresp[['pidp', 'g_istrtdaty', 'g_scghq1_dv']].copy() # wave 7 (no social support variables)
i_mh = i_indresp[['pidp', 'i_istrtdaty', 'i_scghq1_dv']].copy() # wave 9 (no social support variables)
k_mh_ss = k_indresp[['pidp', 'k_istrtdaty', 'k_scghq1_dv', 'k_scfrendany', 'k_scpundstnd', 'k_scprely', 'k_scpopenup', 'k_scpcritic', 'k_scpletdwn', 'k_scpannoy', 'k_screlany', 'k_scrundstnd', 'k_scrrely', 'k_scropenup', 'k_scrcritic', 'k_scrletdwn', 'k_scrannoy', 'k_scfundstnd', 'k_scfrely', 'k_scfopenup', 'k_scfcritic', 'k_scfletdwn', 'k_scfannoy']].copy() # wave 11
m_mh = m_indresp[['pidp', 'm_istrtdaty', 'm_scghq1_dv']].copy() # wave 13 (no social support variables)
n_mh_ss = n_indresp[['pidp', 'n_istrtdaty', 'n_scghq1_dv', 'n_scfrendany', 'n_scpundstnd', 'n_scprely', 'n_scpopenup', 'n_scpcritic', 'n_scpletdwn', 'n_scpannoy', 'n_screlany', 'n_scrundstnd', 'n_scrrely', 'n_scropenup', 'n_scrcritic', 'n_scrletdwn', 'n_scrannoy', 'n_scfundstnd', 'n_scfrely', 'n_scfopenup', 'n_scfcritic', 'n_scfletdwn', 'n_scfannoy']].copy() # wave 14
o_mh = o_indresp[['pidp', 'o_istrtdaty', 'o_scghq1_dv']].copy() # wave 15 (no social support variables)

# Filter out rows with NaN or negative GHQ (scghq1_dv) in each wave
b_mh_ss = b_mh_ss[b_mh_ss['b_scghq1_dv'].notna() & (pd.to_numeric(b_mh_ss['b_scghq1_dv'], errors='coerce') >= 0)].copy()
e_mh_ss = e_mh_ss[e_mh_ss['e_scghq1_dv'].notna() & (pd.to_numeric(e_mh_ss['e_scghq1_dv'], errors='coerce') >= 0)].copy()
g_mh = g_mh[g_mh['g_scghq1_dv'].notna() & (pd.to_numeric(g_mh['g_scghq1_dv'], errors='coerce') >= 0)].copy()
i_mh = i_mh[i_mh['i_scghq1_dv'].notna() & (pd.to_numeric(i_mh['i_scghq1_dv'], errors='coerce') >= 0)].copy()
k_mh_ss = k_mh_ss[k_mh_ss['k_scghq1_dv'].notna() & (pd.to_numeric(k_mh_ss['k_scghq1_dv'], errors='coerce') >= 0)].copy()
m_mh = m_mh[m_mh['m_scghq1_dv'].notna() & (pd.to_numeric(m_mh['m_scghq1_dv'], errors='coerce') >= 0)].copy()
n_mh_ss = n_mh_ss[n_mh_ss['n_scghq1_dv'].notna() & (pd.to_numeric(n_mh_ss['n_scghq1_dv'], errors='coerce') >= 0)].copy()
o_mh = o_mh[o_mh['o_scghq1_dv'].notna() & (pd.to_numeric(o_mh['o_scghq1_dv'], errors='coerce') >= 0)].copy()

# Creating a new column 'positive_spouse_ss' as the average of reverse-coded positive items when >=2 inputs are valid
b_spouse_valid = b_mh_ss[['b_scpundstnd', 'b_scprely', 'b_scpopenup']].where(b_mh_ss[['b_scpundstnd', 'b_scprely', 'b_scpopenup']] >= 0)
e_spouse_valid = e_mh_ss[['e_scpundstnd', 'e_scprely', 'e_scpopenup']].where(e_mh_ss[['e_scpundstnd', 'e_scprely', 'e_scpopenup']] >= 0)
k_spouse_valid = k_mh_ss[['k_scpundstnd', 'k_scprely', 'k_scpopenup']].where(k_mh_ss[['k_scpundstnd', 'k_scprely', 'k_scpopenup']] >= 0)
n_spouse_valid = n_mh_ss[['n_scpundstnd', 'n_scprely', 'n_scpopenup']].where(n_mh_ss[['n_scpundstnd', 'n_scprely', 'n_scpopenup']] >= 0)

b_mh_ss['b_positive_spouse_ss'] = np.where(
    b_spouse_valid.notna().sum(axis=1) >= 2,
    (5 - b_spouse_valid).mean(axis=1),
    np.nan
)
e_mh_ss['e_positive_spouse_ss'] = np.where(
    e_spouse_valid.notna().sum(axis=1) >= 2,
    (5 - e_spouse_valid).mean(axis=1),
    np.nan
)
k_mh_ss['k_positive_spouse_ss'] = np.where(
    k_spouse_valid.notna().sum(axis=1) >= 2,
    (5 - k_spouse_valid).mean(axis=1),
    np.nan
)
n_mh_ss['n_positive_spouse_ss'] = np.where(
    n_spouse_valid.notna().sum(axis=1) >= 2,
    (5 - n_spouse_valid).mean(axis=1),
    np.nan
)

# Removing unnecessary columns to reduce dataframe size
b_mh_ss = b_mh_ss.drop(columns=['b_scpundstnd', 'b_scprely', 'b_scpopenup'])
e_mh_ss = e_mh_ss.drop(columns=['e_scpundstnd', 'e_scprely', 'e_scpopenup'])
k_mh_ss = k_mh_ss.drop(columns=['k_scpundstnd', 'k_scprely', 'k_scpopenup'])
n_mh_ss = n_mh_ss.drop(columns=['n_scpundstnd', 'n_scprely', 'n_scpopenup'])

# Creating a new column 'positive_family_ss' as the average of reverse-coded items when >=2 inputs are valid
b_family_valid = b_mh_ss[['b_scrundstnd', 'b_scrrely', 'b_scropenup']].where(b_mh_ss[['b_scrundstnd', 'b_scrrely', 'b_scropenup']] >= 0)
e_family_valid = e_mh_ss[['e_scrundstnd', 'e_scrrely', 'e_scropenup']].where(e_mh_ss[['e_scrundstnd', 'e_scrrely', 'e_scropenup']] >= 0)
k_family_valid = k_mh_ss[['k_scrundstnd', 'k_scrrely', 'k_scropenup']].where(k_mh_ss[['k_scrundstnd', 'k_scrrely', 'k_scropenup']] >= 0)
n_family_valid = n_mh_ss[['n_scrundstnd', 'n_scrrely', 'n_scropenup']].where(n_mh_ss[['n_scrundstnd', 'n_scrrely', 'n_scropenup']] >= 0)

b_mh_ss['b_positive_family_ss'] = np.where(
    b_family_valid.notna().sum(axis=1) >= 2,
    (5 - b_family_valid).mean(axis=1),
    np.nan
)
e_mh_ss['e_positive_family_ss'] = np.where(
    e_family_valid.notna().sum(axis=1) >= 2,
    (5 - e_family_valid).mean(axis=1),
    np.nan
)
k_mh_ss['k_positive_family_ss'] = np.where(
    k_family_valid.notna().sum(axis=1) >= 2,
    (5 - k_family_valid).mean(axis=1),
    np.nan
)
n_mh_ss['n_positive_family_ss'] = np.where(
    n_family_valid.notna().sum(axis=1) >= 2,
    (5 - n_family_valid).mean(axis=1),
    np.nan
)

# Removing unnecessary columns to reduce dataframe size
b_mh_ss = b_mh_ss.drop(columns=['b_scrundstnd', 'b_scrrely', 'b_scropenup'])
e_mh_ss = e_mh_ss.drop(columns=['e_scrundstnd', 'e_scrrely', 'e_scropenup'])
k_mh_ss = k_mh_ss.drop(columns=['k_scrundstnd', 'k_scrrely', 'k_scropenup'])
n_mh_ss = n_mh_ss.drop(columns=['n_scrundstnd', 'n_scrrely', 'n_scropenup'])

# Creating a new column 'positive_friend_ss' as the average of reverse-coded items when >=2 inputs are valid
b_friend_valid = b_mh_ss[['b_scfundstnd', 'b_scfrely', 'b_scfopenup']].where(b_mh_ss[['b_scfundstnd', 'b_scfrely', 'b_scfopenup']] >= 0)
e_friend_valid = e_mh_ss[['e_scfundstnd', 'e_scfrely', 'e_scfopenup']].where(e_mh_ss[['e_scfundstnd', 'e_scfrely', 'e_scfopenup']] >= 0)
k_friend_valid = k_mh_ss[['k_scfundstnd', 'k_scfrely', 'k_scfopenup']].where(k_mh_ss[['k_scfundstnd', 'k_scfrely', 'k_scfopenup']] >= 0)
n_friend_valid = n_mh_ss[['n_scfundstnd', 'n_scfrely', 'n_scfopenup']].where(n_mh_ss[['n_scfundstnd', 'n_scfrely', 'n_scfopenup']] >= 0)

b_mh_ss['b_positive_friend_ss'] = np.where(
    b_friend_valid.notna().sum(axis=1) >= 2,
    (5 - b_friend_valid).mean(axis=1),
    np.nan
)
e_mh_ss['e_positive_friend_ss'] = np.where(
    e_friend_valid.notna().sum(axis=1) >= 2,
    (5 - e_friend_valid).mean(axis=1),
    np.nan
)
k_mh_ss['k_positive_friend_ss'] = np.where(
    k_friend_valid.notna().sum(axis=1) >= 2,
    (5 - k_friend_valid).mean(axis=1),
    np.nan
)
n_mh_ss['n_positive_friend_ss'] = np.where(
    n_friend_valid.notna().sum(axis=1) >= 2,
    (5 - n_friend_valid).mean(axis=1),
    np.nan
)

# Removing unnecessary columns to reduce dataframe size
b_mh_ss = b_mh_ss.drop(columns=['b_scfundstnd', 'b_scfrely', 'b_scfopenup'])
e_mh_ss = e_mh_ss.drop(columns=['e_scfundstnd', 'e_scfrely', 'e_scfopenup'])
k_mh_ss = k_mh_ss.drop(columns=['k_scfundstnd', 'k_scfrely', 'k_scfopenup'])
n_mh_ss = n_mh_ss.drop(columns=['n_scfundstnd', 'n_scfrely', 'n_scfopenup'])
```
